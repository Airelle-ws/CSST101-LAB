**3. Numerical Operators**

```r
x <- 10
y <- 3

add <- x + y
sub <- x - y
mul <- x * y
div <- x / y
mod <- x %% y
exp <- x ^ y

print(add)
print(sub)
print(mul)
print(div)
print(mod)
print(exp)
```

**8. Vectors**

```r
grades <- c(98,80,91,88,96)

grades

cat("MEAN: ",mean(grades))
cat("\nMax: ",max(grades))
cat("\nMin: ",min(grades))
cat("\nLength: ",length(grades))
cat("\nMedian: ",median(grades))
summary(grades)
```
