2. Basic **Caculation**


```r
# Student grades
quiz <- 85
laboratory <- 90
project <- 88
exam <- 80

# Grading weights
quiz_weight <- 0.20
laboratory_weight <- 0.25
project_weight <- 0.25
exam_weight <- 0.30

# Calculate weighted final grade
final_grade <- (quiz * quiz_weight) +
               (laboratory * laboratory_weight) +
               (project * project_weight) +
               (exam * exam_weight)

# Display the final grade
print(final_grade)
```

**7. Creating a Function**

```r
classify_grade <- function(grade) {

  if (grade >= 90 && grade <= 100) {
    return("Excellent")

  } else if (grade >= 85 && grade <= 89) {
    return("Very Good")

  } else if (grade >= 80 && grade <= 84) {
    return("Good")

  } else if (grade >= 75 && grade <= 79) {
    return("Passed")

  } else {
    return("Failed")
  }
}

# Example
grade <- 74
print(classify_grade(grade))
```

**Creating a Simple Rule-Based System**

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)

students$Risk <- ifelse(
  students$Attendance < 75 & students$Grade < 75,
  "High Risk",
  "Low Risk"
)

print(students)
```

**11 & 12. Apply the Rule to Multiple Students**

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)

student_risk <- function(attendance, grade) {
  if (attendance < 75 && grade < 75) {
    return("High Risk")
  } else {
    return("Low Risk")
  }
}

students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

students
```

**13. Create a Visualization**

```r
# Create the student data
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)

# Create the risk rule
student_risk <- function(attendance, grade) {
  if (attendance < 75 && grade < 75) {
    return("High Risk")
  } else {
    return("Low Risk")
  }
}

# Apply the rule
students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

# Count the risk categories
risk_counts <- table(students$Risk)

# Create a bar chart
barplot(
  risk_counts,
  main = "Student Risk Classification",
  xlab = "Risk Level",
  ylab = "Number of Students"
)
```

Answers:
What does the graph show?
→ It shows the number of students in each risk category.
Which risk category has the most students?
→ Low Risk, with 3 students.
What action should the university take?
→ The university should give attention and support to High Risk students, especially those with low attendance and grades.

**14. Add Recommendation**

```r
# Create the student data
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)

# Create the risk rule
student_risk <- function(attendance, grade) {
  if (attendance < 75 && grade < 75) {
    return("High Risk")
  } else {
    return("Low Risk")
  }
}

# Apply the risk rule
students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

# Add recommendation
students$Recommendation <- ifelse(
  students$Risk == "High Risk",
  "Immediate academic intervention",
  "Continue regular monitoring"
)

# Display the result
print(students)
```
